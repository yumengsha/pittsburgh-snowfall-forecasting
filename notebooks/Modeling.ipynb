```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/processed/snow_removal_feature_engineered.csv')

# Leakage-safe feature rebuild.
# Every request-history feature is computed per neighborhood and shifted by one row,
# so a row only sees requests that were already known before that day.
# Weather features (SNOW, TAVG, TMAX) are treated as same-day forecasts and are not shifted.
# Note: the table only has rows for neighborhood-days with at least one request,
# so "lag_1d" / "rolling_mean_3d" refer to the previous observed request-days of that neighborhood.

# 1. Make sure 'date' is datetime and sort chronologically within each neighborhood
df['date'] = pd.to_datetime(df['date'])
df = df.sort_values(['neighborhood', 'date'])
g = df.groupby('neighborhood')

# 2. Lag features (past values only)
df['lag_1d'] = g['daily_request_count'].shift(1)
df['lag_3d'] = g['daily_request_count'].shift(3)

# 3. Rolling features
# shift(1) BEFORE rolling: the window excludes the same-day target
df['rolling_mean_3d'] = g['daily_request_count'].transform(lambda s: s.shift(1).rolling(3).mean())
df['rolling_sum_snow_3d'] = g['SNOW'].transform(lambda s: s.rolling(3).sum())

# 4. Calendar features
df['month'] = df['date'].dt.month
df['dayofweek'] = df['date'].dt.dayofweek  # 0=Monday, ..., 6=Sunday
df['is_weekend'] = df['dayofweek'] >= 5

# 5. Final selected columns
selected_cols_expanded = [
    'daily_request_count',   # target
    'SNOW',
     'TAVG',
    'TMAX',
    'lag_1d',
     'rolling_mean_3d',
    'rolling_sum_snow_3d',
     'month',
     'is_snow_season'
]

# 6. Create final feature dataset (date index, used for the chronological split)
data = df.set_index('date')[selected_cols_expanded].fillna(0).copy()

print("✅ Feature expansion completed. Shape:", data.shape)
data.head()
```

```python
# 1. Import necessary libraries
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import lightgbm as lgb
import pandas as pd
import numpy as np

# 2. Chronological train-test split (train: <= 2022, holdout: 2023+)
train = data[data.index.year <= 2022]
test = data[data.index.year >= 2023]

# Separate target and features
y_train = train['daily_request_count']
X_train = train.drop(columns=['daily_request_count']).astype(float)

y_test = test['daily_request_count']
X_test = test.drop(columns=['daily_request_count']).astype(float)

# 3. Build and Train LightGBM Model
lgbm_model = lgb.LGBMRegressor(
    objective='regression',
    boosting_type='gbdt',
    learning_rate=0.01,
    n_estimators=1000,
    max_depth=5,
    num_leaves=15,                # reduce leaves
    subsample=0.8,                
    colsample_bytree=0.8,
    reg_alpha=1.0,                # L1 regularization (adds sparsity)
    reg_lambda=2.0,               # L2 regularization (smooths weights)
    min_child_samples=50,         # minimum samples in one leaf
    random_state=42,
    verbose=-1
)


# Timing training
import time
start = time.time()

lgbm_model.fit(X_train, y_train)

end = time.time()
print(f"✅ Model trained in {end - start:.2f} seconds.")

# 4. Predict
y_train_pred = lgbm_model.predict(X_train)
y_test_pred = lgbm_model.predict(X_test)

# 5. Evaluate
mae_train = mean_absolute_error(y_train, y_train_pred)
rmse_train = np.sqrt(mean_squared_error(y_train, y_train_pred))
r2_train = r2_score(y_train, y_train_pred)
mape_train = np.mean(np.abs((y_train - y_train_pred) / np.maximum(1, y_train))) * 100

mae_test = mean_absolute_error(y_test, y_test_pred)
rmse_test = np.sqrt(mean_squared_error(y_test, y_test_pred))
r2_test = r2_score(y_test, y_test_pred)
mape_test = np.mean(np.abs((y_test - y_test_pred) / np.maximum(1, y_test))) * 100

# 6. Display Results
metrics_df = pd.DataFrame({
    'Metric': ['MAE', 'RMSE', 'MAPE (%)', 'R-squared (R²)'],
    'Train': [mae_train, rmse_train, mape_train, r2_train],
    'Test': [mae_test, rmse_test, mape_test, r2_test]
})

print("\n📊 Performance Metrics (Expanded Features):")
print(metrics_df.to_string(index=False))

# 7. Baselines on the same 2023+ holdout
baselines = {
    'LightGBM': y_test_pred,
    'Lag-1 (previous request-day)': X_test['lag_1d'].values,
    'Train mean': np.full(len(y_test), y_train.mean()),
}
baseline_df = pd.DataFrame([
    {'Model': name,
     'MAE': mean_absolute_error(y_test, pred),
     'RMSE': np.sqrt(mean_squared_error(y_test, pred))}
    for name, pred in baselines.items()
])

print("\n📊 Holdout (2023+) vs. Baselines:")
print(baseline_df.to_string(index=False, float_format='%.2f'))
```
